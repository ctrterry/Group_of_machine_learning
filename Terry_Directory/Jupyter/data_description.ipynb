```python
import pandas as pd

# Define IMDb rating bins and labels
bins = [0, 5, 6, 7, 8, 10]
labels = ['Below 5', '5 to 6', '6 to 7', '7 to 8', '8 and above']

# Create a new column for the rating bin
read_used['rating_range'] = pd.cut(read_used['averageRating'], bins=bins, labels=labels, right=False)

# Group by rating range and calculate mean actor likes
summary = (
    read_used.groupby('rating_range')[['actor_1_facebook_likes', 'actor_2_facebook_likes', 'actor_3_facebook_likes']]
    .mean()
    .round(0)  #  round to 0 decimals for neatness
    .reset_index()
)

print(summary)
```

```python
####

file_path = '/Users/terrychen/Desktop/Group_of_machine_learning/Terry_Directory/Data/Raw/Raw_Merge_Kaggle_and_Social_Metadata.csv'
budget_and_budget_adj = pd.read_csv(file_path)
```

```python
budget_and_budget_adj.columns
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(14, 8))

# Set a reasonable max budget for the plot (e.g., $200 million)
max_budget = 2e8

# Filter out extreme outliers for the plot
budget_orig = budget_and_budget_adj['budget'].dropna()
budget_adj = budget_and_budget_adj['budget_adjusted'].dropna()
budget_orig = budget_orig[budget_orig <= max_budget]
budget_adj = budget_adj[budget_adj <= max_budget]

# Plot histograms
plt.hist(budget_orig, bins=30, color='blue', alpha=0.4, label='Original Budgets', edgecolor='black')
plt.hist(budget_adj, bins=30, color='red', alpha=0.4, label='CPI-Adjusted Budgets', edgecolor='black')

plt.title('Budget Distribution Before and After CPI Adjustment', fontsize=20)
plt.xlabel('Budget (USD)', fontsize=16)
plt.ylabel('Frequency', fontsize=16)
plt.legend(fontsize=14)
plt.tight_layout()
plt.show()
```
