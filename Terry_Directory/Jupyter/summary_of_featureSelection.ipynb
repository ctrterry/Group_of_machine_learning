### Why do we remove these features?

ID, IMDb ID, tconst: These are unique identifiers and do not provide any predictive value for the model.

Belongs to Collection, Homepage, Poster Path: These are metadata or links related to marketing or UI display, not useful for modeling.

Original Language: Since we're only focusing on English-language movies, this column has no variance and adds no value.

Overview, Tagline, Title, Keywords: These are textual features. While potentially useful with NLP techniques, they require complex preprocessing and are excluded in this version for simplicity.

Budget, Budget Adjusted, Revenue: These features are often missing or inaccurately reported. To avoid introducing noise or relying on unreliable data, we remove them here.

Release Date, Release Day, Status, Spoken Languages: Either redundant, too granular (day vs. date), or non-informative for rating prediction.

Production Companies, Production Countries: These involve many unique values, which could lead to high-cardinality categorical variables. We exclude them to simplify the model.

Cast, Crew: These are complex nested JSON-like structures. Instead, we extract and process only key members (e.g., top actors or director) separately if needed.

Director Name, Actor1, 2, and 3 Name: Names are high-cardinality features and not directly useful unless we extract meaningful numerical or categorical representations (like Facebook likes or prior success). For this model, we exclude raw name fields.
